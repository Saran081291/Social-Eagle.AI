# 4. Python Virtual Environment

A **virtual environment** is a *separate, private space* for each Python project.

It keeps the libraries (extra tools) of one project away from another project.

## Why do we need it? (Simple example)

Imagine:
- **Project A** needs version **1.0** of a tool.
- **Project B** needs version **2.0** of the same tool.

If everything is installed in one place, they will clash.

A virtual environment gives **each project its own box**, so there is no conflict.

## How to create a virtual environment

These commands are typed in the **terminal** (not inside Python).

> Note: lines starting with `#` below are explanations, not commands.

### Step 1: Create the environment
```
python -m venv myenv
```
This creates a folder named `myenv` that holds the private environment.

### Step 2: Activate it

**On Windows:**
```
myenv\Scripts\activate
```

**On Mac / Linux:**
```
source myenv/bin/activate
```

When active, you will see `(myenv)` at the start of the terminal line.

### Step 3: Install a library inside it
```
pip install requests
```

### Step 4: Deactivate when done
```
deactivate
```

## Check your Python version (run this in a cell)

The code below shows which Python is being used.

````python
import sys
print("Python version:", sys.version)
print("Python location:", sys.executable)
````

## Key points to remember

- A virtual environment is a private space for one project.
- It avoids version clashes between projects.
- Create: `python -m venv myenv`
- Activate (Windows): `myenv\Scripts\activate`
- Deactivate: `deactivate`
