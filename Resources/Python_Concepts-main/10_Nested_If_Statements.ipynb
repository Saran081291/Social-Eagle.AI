# 10. Nested If Statements

A **nested if** means putting one `if` statement **inside another `if`**.

We use it when we need to check a **second condition** only after the first one is True.

## Real-life example

- **IF** you have money:
  - **IF** the shop is open → buy items.
  - **ELSE** → go back home.

The second check happens **only if** the first is true.

## Example 1: Checking eligibility

The inner `if` is indented further (more spaces) to show it is inside the outer `if`.

```python
thiru = 50
shop = 0

if thiru == 50:
    if shop == 1:
        print("order pongal")

    else:
        print("search for another hotel")
else:
    print("go to home")
```

## Example 2: Largest of two numbers

First we check if both numbers are positive, then we compare them.

```python
a = 8
b = 5

if a > 0 and b > 0:
    if a > b:
        print("a is the largest")
    else:
        print("b is the largest")
else:
    print("Please use positive numbers")
```

## Example 3: Login check

First check the username, then check the password.

```python
username = "admin"
password = "1234"

if username == "admin":
    if password == "1234":
        print("Login successful!")
    else:
        print("Wrong password")
else:
    print("Unknown user")
```

## Key points to remember

- A nested if is an `if` inside another `if`.
- The inner check runs **only when** the outer condition is True.
- Each inner block needs **extra indentation** (more spaces).
- Don't nest too deeply — it becomes hard to read.

```python
def calculator():
    try:
        num1 = float(input("Enter first number: "))
        operator = input("Enter operator (+, -, *, /): ")
        num2 = float(input("Enter second number: "))

        if operator == "+":
            result = num1 + num2
        elif operator == "-":
            result = num1 - num2
        elif operator == "*":
            result = num1 * num2
        elif operator == "/":
            if num2 == 0:
                raise ZeroDivisionError("Cannot divide by zero")
            result = num1 / num2
        else:
            raise ValueError("Invalid operator")

        print(f"Result: {result}")

    except Exception as e:
        print(f"Error: {e}")

calculator()
```
