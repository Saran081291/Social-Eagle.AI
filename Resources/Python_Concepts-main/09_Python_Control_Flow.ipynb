# 9. Python Control Flow

**Control flow** means making decisions in our program.

The program chooses *which code to run* based on a **condition** (something that is True or False).

We use: `if`, `elif`, and `else`.

## Real-life example

- **IF** it is raining → take an umbrella.
- **ELSE** → wear sunglasses.

This is exactly how `if` and `else` work.

## 1. The `if` statement

Runs the code **only if** the condition is True.

👉 Note the **colon `:`** and the **indentation** (spaces).

```python
age = 20

if age >= 18:
    print("You are an adult.")
```

## 2. The `if ... else` statement

If the condition is True → run the `if` block.

Otherwise → run the `else` block.

```python
age = 15

if age >= 18:
    print("You can vote.")
else:
    print("You are too young to vote.")
```

## 3. The `if ... elif ... else` statement

Use `elif` (else-if) to check **many conditions** one by one.

Example: giving a grade based on marks.

```python
thiru = 5

if thiru >= 50:
    print("order pongal")

elif thiru == 30:
    print("order idly")

elif thiru == 10:
    print("order vada")

else:
    print("Go home")
```

## Try it yourself

Change the value of `temperature` and run the cell to see different results.

```python
temperature = 35

if temperature > 30:
    print("It is hot today.")
else:
    print("The weather is pleasant.")
```

## Key points to remember

- `if` runs code when a condition is True.
- `else` runs when the condition is False.
- `elif` checks extra conditions in between.
- Always end the condition line with `:` and **indent** the next line.
