# 13. Error Handling

Sometimes our program meets a problem and **crashes**. This is called an **error** (or exception).

**Error handling** lets us catch these problems and show a friendly message instead of crashing.

## What does an error look like?

The code below tries to divide by zero, which is not allowed.

(Run it to see the error message — this is normal!)

```python
# This will cause an error
result = 10 / 0
print(result)
```

## The `try ... except` block

- **`try`** → write the code that *might* cause an error.
- **`except`** → write what to do *if* an error happens.

This way the program does not crash.

```python
try:
    result = 10 / 0
    print(result)
except:
    print("Oops! You cannot divide by zero.")
```

## Catching a specific error

We can name the error type to handle it clearly.

```python
try:
    number = int("hello")   # cannot turn text 'hello' into a number
except ValueError:
    print("That is not a valid number.")
```

## Using `else` and `finally`

- **`else`** → runs if there was **no** error.
- **`finally`** → runs **always** (error or not).

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Success! Result is:", result)
finally:
    print("This always runs.")
```

## A practical example

Safely asking for a number (here we use a fixed value instead of input).

```python
user_value = "abc"   # imagine the user typed this

try:
    age = int(user_value)
    print("Your age is", age)
except ValueError:
    print("Please enter numbers only!")
```

```python
try:
    number = int("hello")
except Exception as e:
    print(f"An error occurred: {e}")
```

## Key points to remember

- Errors normally crash the program.
- Put risky code in **`try`** and handle problems in **`except`**.
- This keeps the program running smoothly.
- **`else`** runs when there is no error; **`finally`** always runs.
