# 1. What is Programming?

Programming is the process of **giving instructions to a computer** to perform a task.

A computer does not think on its own. It only does what we tell it to do. We tell it using a **programming language** (like Python).

### A simple real-life example

Think of a recipe for making tea:
1. Boil water
2. Add tea powder
3. Add milk and sugar
4. Serve

This step-by-step list of instructions is exactly what a **program** is — but written for a computer.

## Why do we need programming?

We use programs every day:
- **WhatsApp** to send messages
- **YouTube** to watch videos
- **Google Maps** to find directions
- **Calculator** to do math

All of these are built using programming.

## What is a programming language?

Humans speak languages like English, Hindi, or Tamil.

Computers understand only **0s and 1s** (binary). It is very hard for humans to write that.

So we use **programming languages** (like Python, Java, C) that are easy for humans to read and write. A special tool then converts our code into something the computer understands.
