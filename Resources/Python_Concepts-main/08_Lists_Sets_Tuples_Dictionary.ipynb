# 8. Lists, Sets, Tuples & Dictionary

These are **collections** — they store *many values* in a single variable.

| Type | Symbol | Ordered? | Changeable? | Duplicates? |
|------|--------|----------|-------------|-------------|
| **List** | `[ ]` | Yes | Yes | Yes |
| **Tuple** | `( )` | Yes | No | Yes |
| **Set** | `{ }` | No | Yes | No |
| **Dictionary** | `{key: value}` | Yes | Yes | No (keys) |

## 1. List `[ ]`

A **list** stores multiple items in order. You **can change** it.

Think of a shopping list.

```python
fruits = ["apple", "banana", "mango"]

print(fruits)
print(fruits[0])      # first item (counting starts at 0)

fruits.append("orange")   # add a new item
print(fruits)

fruits[1] = "grapes"      # change an item
print(fruits)
```

## 2. Tuple `( )`

A **tuple** is like a list, but you **cannot change** it (it is fixed).

Good for data that should not change, like days of the week.

```python
colors = ("red", "green", "blue")

print(colors)
print(colors[0])   # we can read items

#colors[0] = "yellow"  # ❌ This would give an ERROR (tuples cannot change)
```

## 3. Set `{ }`

A **set** stores items with **no duplicates** and **no order**.

Useful for keeping only unique values.

```python
numbers = {1, 2, 2, 3, 3, 3}

print(numbers)   # duplicates are automatically removed

numbers.add(4)   # add a new item
print(numbers)
```

## 4. Dictionary `{key: value}`

A **dictionary** stores data as **key–value pairs**.

Like a real dictionary: a word (key) and its meaning (value).

```python
student = {
    "name": "Ravi",
    "age": 20,
    "city": "Delhi"
}

print(student)
print(student["name"])   # get value using its key

student["age"] = 21       # change a value
print(student)

student["grade"] = "A"     # add a new key-value pair
print(student)
```

## Key points to remember

- **List** `[ ]` → ordered, changeable, allows duplicates.
- **Tuple** `( )` → ordered, **fixed** (cannot change).
- **Set** `{ }` → no duplicates, no order.
- **Dictionary** `{key: value}` → stores key–value pairs.
- Counting (indexing) always starts from **0**.
