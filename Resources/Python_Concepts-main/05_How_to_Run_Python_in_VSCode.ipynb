# 5. How to Run Python in VS Code

**VS Code (Visual Studio Code)** is a free, popular code editor where we write and run our Python programs.

## Step-by-step setup

### Step 1: Install Python
- Go to **https://www.python.org/downloads**
- Download and install Python.
- ✅ During install, **tick the box** "Add Python to PATH".

### Step 2: Install VS Code
- Go to **https://code.visualstudio.com**
- Download and install it.

### Step 3: Install the Python Extension
- Open VS Code.
- Click the **Extensions** icon on the left (four squares).
- Search for **"Python"** (by Microsoft) and click **Install**.

## How to run a Python file (.py)

### Step 1: Create a file
- Create a new file and name it `hello.py`

### Step 2: Write code
```python
print("Hello from VS Code!")
```

### Step 3: Run it
- Click the **▶ Run** button (top-right corner), **OR**
- Right-click the file → **Run Python File in Terminal**

The output appears in the **terminal** at the bottom.

## Running code in a Jupyter Notebook (.ipynb)

This file you are reading is a **notebook**. To run a cell:
- Click on the cell
- Press **Shift + Enter**

## Key points to remember

- Install **Python**, then **VS Code**, then the **Python extension**.
- `.py` files run with the **▶ Run** button.
- In notebooks (`.ipynb`), run a cell with **Shift + Enter**.
- Output appears in the terminal (or below the cell).
