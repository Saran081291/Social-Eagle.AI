# 12. Python Functions

A **function** is a block of code that does a specific task.

We write it **once** and can **use it many times**.

This keeps our code short and clean.

## Real-life example

A **washing machine** is like a function:
- You give it clothes (**input**)
- It does the work (**task**)
- It gives back clean clothes (**output**)

## 1. Creating and calling a function

- We **define** a function using the `def` keyword.
- We **call** (use) it by writing its name with `()`.

```python
def add(a,b,c):
    c = a+b+c
    return c

add(2,3,4)
```

## 2. Function with parameters (inputs)

We can pass values into a function. These are called **parameters**.

```python
def greet(name):
    print("Hello,", name)

greet("Ravi")
greet("Anjali")
```

## 3. Function with `return` (output)

`return` sends a value **back** so we can use it later.

```python
def add(a, b):
    return a + b

result = add(10, 5)   # store the returned value
print("Sum is:", result)
```

## 4. A complete example

A function that calculates the area of a rectangle.

```python
def area_of_rectangle(length, width):
    area = length * width
    return area

print("Area:", area_of_rectangle(5, 3))
print("Area:", area_of_rectangle(10, 2))
```

## Key points to remember

- A function is reusable code that does one task.
- Create it with `def function_name():`.
- **Parameters** are inputs you pass in `( )`.
- **`return`** gives a value back from the function.
- Call a function by writing its name with `()`.
