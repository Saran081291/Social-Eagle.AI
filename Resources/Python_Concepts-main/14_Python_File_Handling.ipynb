# 14. Python File Handling

**File handling** means working with files — **creating**, **writing**, and **reading** them using Python.

This lets our program **save data** so it is not lost when the program closes.

## File modes

When we open a file, we tell Python what we want to do:

| Mode | Meaning |
|------|---------|
| `"w"` | Write (creates a new file or overwrites it) |
| `"r"` | Read (read the content) |
| `"a"` | Append (add to the end of the file) |

## 1. Writing to a file

This creates a file named `myfile.txt` and writes text into it.

Using `with` is the recommended way — it closes the file automatically.

```python
with open("myfile.txt", "w") as file:
    file.write("Hello social eagle\n")

print("File written successfully!")
```

## 2. Reading from a file

Now let's read back what we wrote.

```python
with open("myfile.txt", "r") as file:
    content = file.read()

print(content)
```

## 3. Appending to a file

`"a"` mode adds new text to the **end** without deleting the old text.

```python
with open("myfile.txt", "a") as file:
    file.write("\nThis line was added later.")

print("Text added!")
```

## 4. Reading line by line

We can loop through a file to read one line at a time.

```python
with open("myfile.txt", "r") as file:
    for line in file:
        print(line.strip())   # strip() removes extra spaces/newlines
```

## Key points to remember

- File handling saves data permanently.
- Use `open("filename", mode)` to open a file.
- Modes: `"w"` write, `"r"` read, `"a"` append.
- Use `with open(...) as file:` — it closes the file automatically.
- `.write()` writes text, `.read()` reads text.
