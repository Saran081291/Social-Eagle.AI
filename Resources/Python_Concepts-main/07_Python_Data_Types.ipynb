# 7. Python Data Types

A **data type** tells us *what kind of value* a variable holds.

For example: a number, some text, or a True/False value.

## What is a variable?

A **variable** is like a box with a name that stores a value.

```python
name = "Anjali"   # storing text in a box named 'name'
age = 21          # storing a number
print(name)
print(age)
```

## Main data types in Python

| Type | Name | Example |
|------|------|---------|
| `int` | Integer (whole number) | `10`, `-5` |
| `float` | Decimal number | `3.14`, `2.5` |
| `str` | String (text) | `"Hello"` |
| `bool` | Boolean (True/False) | `True`, `False` |

## 1. Integer (int) — whole numbers

```python
marks = 95
print(marks)
print(type(marks))   # type() shows the data type
```

## 2. Float (float) — decimal numbers

```python
price = 49.99
print(price)
print(type(price))
```

## 3. String (str) — text

Any value inside quotes `" "` is a string.

```python
city = 90
print(city)
print(type(city))
```

## 4. Boolean (bool) — True or False

```python
is_student = False
print(is_student)
print(type(is_student))
```

## Converting between types

We can change one type into another using `int()`, `float()`, `str()`.

```python
number_text = "100"          # this is a string
number = int(number_text)    # convert text to integer
print(number + 50)            # now we can do math
```

## Key points to remember

- A variable stores a value (like a named box).
- `int` = whole numbers, `float` = decimals.
- `str` = text (inside quotes), `bool` = True/False.
- Use `type()` to check the data type.
- Use `int()`, `float()`, `str()` to convert types.
