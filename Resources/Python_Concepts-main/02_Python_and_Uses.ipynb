# 2. Python & Its Uses

**Python** is a popular programming language created by **Guido van Rossum** in 1991.

It is famous because it is:
- **Easy to read** (looks almost like English)
- **Simple to learn** (great for beginners)
- **Powerful** (used by big companies like Google, Netflix, Instagram)

## A small example of Python's power

With just a few lines, Python can do useful work — like calculating the average of marks.

```python
marks = [80, 90, 70, 100]
average = sum(marks) / len(marks)
print("Average marks:", average)
```
