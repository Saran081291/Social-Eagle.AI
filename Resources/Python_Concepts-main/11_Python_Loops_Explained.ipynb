# 11. Python Loops Explained

A **loop** is used to **repeat** a block of code many times.

Instead of writing the same line 100 times, we use a loop.

Python has two main loops: **`for`** and **`while`**.

## 1. The `for` loop

Used to repeat code a **fixed number of times** or to go through a list.

### Using `range()` to repeat

`range(5)` gives numbers `0, 1, 2, 3, 4`.

```python
for i in range(1,5):
    print("Number:", i)
```

### Looping through a list

We can go through each item in a list one by one.

```python
fruits = ["apple", "banana", "mango"]

for thiru in fruits:
    print("I like", thiru)
```

## 2. The `while` loop

Repeats code **as long as a condition is True**.

⚠️ Remember to update the variable, or the loop will run forever!

```python
count = 1

while count <= 5:
    print("Count is:", count)
    count = count + 1   # increase count, so the loop can stop
```

## 3. `break` and `continue`

- **`break`** → stops the loop completely.
- **`continue`** → skips the current turn and moves to the next.

```python
# break example: stop when number is 3
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

```python
# continue example: skip number 3
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

## A practical example: sum of numbers 1 to 5

```python
total = 0

for number in range(1, 6):
    total = total + number

print("Total sum is:", total)
```

## Key points to remember

- Loops repeat code so we don't write it again and again.
- **`for`** loop → for a known number of times / lists.
- **`while`** loop → repeats while a condition is True.
- **`break`** stops the loop; **`continue`** skips one turn.
- `range(n)` gives numbers from 0 up to n-1.
